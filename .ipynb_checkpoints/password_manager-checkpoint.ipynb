```python
class Password:
    def __init__(self):
        self.__user_data={}
        self.__faculty_data={}
    def __new_user(self):
        user_name=input("enter your name.")
        password=input(f"enter password:")
        self.__user_data.update({user_name:password})
    def __edit(self):
        user_name=input("enter the user name")
        password=input("enter the password:")
        if(self.__user_data[user_name]==password):
            choice=int(input("enter 0 to edit the user_name\nenter 1 to edit the change the password:"))
            if(choice==1):
                new_password=int("input your new password:")
                self.__user_data.update({user_name:new_password})
            elif(choice==0):
                new_user_name-input("enter new user name")
                self.__user_data.update({new_user_name:password})
    def __close(self):
        user_name=input("enter the user name which account you want to close")
        password=input("enter the password:")
        if(self.__user_data[user_name]==password):
            del self.__user_data["user_name"]
    def __add_employee(self):
        name=input("enter your name sir...")
        password=input("set your  password sir")
        self.__faculty_data.update({name:password})
    def __analyse_user_data(self,name):
        faculty_name=input("enter your name sir")
        password=input("enter your password")
        if(self.__faculty_data[faculty_name]==password):
            print(f"the user name {name} having password {self.__user_data[name]}")
    def show_user_data(self):
        name=input("enter the name of the the user:")
        password=input("enter the password for your name")
        if(self.__user_data[name]==password):
            print("the data for the user is shown below")
```
