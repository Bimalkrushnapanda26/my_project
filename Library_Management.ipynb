```python
books = {}
import time
op = {}
study_hour = {}
book_status = {}
administrator = {1:"123"}
student = {}
identity = input("Login as Student or Administrator: ")
if identity == "Administrator":
    admin_id = int(input("Enter your ID as admin: "))
    if admin_id in administrator:
        i = 1
        while i <= 3:
            password = input("Enter your password: ")
            if password == administrator[admin_id]:
                while True:
                    n = int(input(
                        "Enter 1 for book allocation\n"
                        "Enter 2 for delete any book from the library\n"
                        "Enter 3 for allocate new version of any book\n"
                        "Enter 4 for displaying the available books\n"
                        "Enter 5 to exit\n"
                    ))
                    if n == 1:
                        book_name = input("Enter the name of the book: ")
                        book_author = input("Enter the author of the book: ")
                        books.update({book_name: book_author})
                    elif n == 2:
                        book_name = input("Enter the book name which you want to delete: ")
                        if book_name in books:
                            del books[book_name]
                        else:
                            print("Book not found.")
                    elif n == 3:
                        book_name = input("Enter the book name which you want to update: ")
                        if book_name in books:
                            del books[book_name]
                            updated_book_name = input("Enter the updated book name: ")
                            updated_book_author_name = input("Enter the author of updated book: ")
                            books.update({updated_book_name: updated_book_author_name})
                        else:
                            print("Book not found.")
                    elif n == 4:
                        for key, value in books.items():
                            print(f"{key} : {value}")
                    elif n == 5:
                        print("Thank you for visiting Library server sir...")
                        break
                    else:
                        print("Read the instruction correctly.")
                    print("\n")
                break
            else:
                print("Incorrect password.")
                i += 1
    else:
        print("Administrator ID not found.")
else:
    while True:
        n = int(input(
            "Enter 1 for taking book from library\n"
            "Enter 2 for giving back the library book\n"
            "Enter 3 for reading in the library\n"
            "Enter 4 for coming back to hostel\n"
        ))
        if n == 1:
            student_roll = input("Enter your roll id: ")
            book_name = input("Enter the name of the book: ")
            if book_name in books:
                op.update({student_roll: book_name})
                book_status.update({student_roll: time.time()})
            else:
                print("Sorry, the book is not available.")
        elif n == 2:
            student_roll = input("Enter your roll id: ")
            if student_roll in op:
                book_status[student_roll] = "Not taken"
                del op[student_roll]
            else:
                print("No book record found.")
        elif n == 3:
            student_roll = input("Enter your roll id: ")
            entry_time = time.time()
            study_hour[student_roll] = entry_time
        elif n == 4:
            student_roll = input("Enter your roll id: ")
            exit_time = time.time()
            if student_roll in study_hour:
                total_time = exit_time - study_hour[student_roll]
                study_hour[student_roll]=total_time
                print(f"Total study time: {total_time} seconds")
            else:
                print(f"Call the student with student id: {student_roll}")
            break
        else:
            print("Invalid option.")
```
